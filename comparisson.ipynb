Modelos predictores de precio de casa

```python
import pandas as pd
import matplotlib.pyplot as plt

# Crear DataFrame con los resultados
data = {
    'Modelo': [
        'MLR (Multivariada)',
        'SVM Lineal',
        'Árbol Regresión (depth=6)',
        'Naive Bayes',
        'KNN',
        'Red Neuronal (ReLU)'
    ],
    'MAE': [
        23967.658177,
        11374.91,
        26487.921,
        27396.6141,
        20596.2566,
        23561.09
    ],
    'RMSE': [
        38093.86,
        20106.69,
        38951.399,
        48012.18,
        33341.10,
        35576.25
    ]
}

df_modelos = pd.DataFrame(data)

# Mostrar tabla
print(df_modelos)

# Gráfico de barras agrupadas (MAE, RMSE, R2)
df_plot = df_modelos.set_index('Modelo')[['MAE', 'RMSE']]
df_plot.plot(kind='bar', figsize=(12, 6))
plt.title('Comparación de Modelos de Predicción de Precio de Casa')
plt.ylabel('Valor')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.legend(title='Métrica')
plt.show()
```

Tiempo y R^2

```python
import pandas as pd
import matplotlib.pyplot as plt

# Datos
data = {
    'Modelo': [
        'MLR (Multivariada)',
        'SVM Lineal',
        'Árbol Regresión (depth=6)',
        'Naive Bayes',
        'KNN',
        'Red Neuronal (ReLU)'
    ],
    'R2': [
        0.818402,
        0.9473,
        0.807,
        0.6605,
        0.8363,
        0.8350
    ],
    'Tiempo (s)': [
        0.3,
        1.6,
        0.4,
        20.0,
        0.1,
        107
    ]
}

df = pd.DataFrame(data)

# Gráfico 1: R² por modelo
plt.figure(figsize=(10, 5))
plt.bar(df['Modelo'], df['R2'], color='skyblue')
plt.title('Comparación de R² por Modelo')
plt.ylabel('R²')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# Gráfico 2: Tiempo de ejecución por modelo
plt.figure(figsize=(10, 5))
plt.bar(df['Modelo'], df['Tiempo (s)'], color='salmon')
plt.title('Comparación de Tiempo de Ejecución por Modelo')
plt.ylabel('Tiempo (segundos)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
