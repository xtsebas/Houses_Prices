# Modelos predictores de precio de casa

```python
import pandas as pd
import matplotlib.pyplot as plt

# Crear DataFrame con los resultados
data = {
    'Modelo': [
        'MLR (Multivariada)',
        'SVM Lineal',
        'Árbol Regresión (depth=6)',
        'Naive Bayes',
        'KNN',
        'Red Neuronal (ReLU)'
    ],
    'MAE': [
        23967.658177,
        11374.91,
        26487.921,
        27396.6141,
        20596.2566,
        23561.09
    ],
    'RMSE': [
        38093.86,
        20106.69,
        38951.399,
        48012.18,
        33341.10,
        35576.25
    ]
}

df_modelos = pd.DataFrame(data)

# Mostrar tabla
print(df_modelos)

# Gráfico de barras agrupadas (MAE, RMSE, R2)
df_plot = df_modelos.set_index('Modelo')[['MAE', 'RMSE']]
df_plot.plot(kind='bar', figsize=(12, 6))
plt.title('Comparación de Modelos de Predicción de Precio de Casa')
plt.ylabel('Valor')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.legend(title='Métrica')
plt.show()
```

Conclusión Comparativa de Modelos

El modelo SVM Lineal obtuvo el MAE más bajo (11,374.91) y el RMSE más bajo (20,106.69), superando a todos los demás en precisión absoluta y cuadrática.

KNN también mostró un desempeño sólido, con el segundo mejor MAE (20,596.26) y RMSE (33,341.10), seguido de la Red Neuronal y MLR.

Naive Bayes y el Árbol de Regresión presentaron los errores más altos, siendo menos confiables para predicciones precisas.

Tiempo y R^2

```python
import pandas as pd
import matplotlib.pyplot as plt

# Datos
data = {
    'Modelo': [
        'MLR (Multivariada)',
        'SVM Lineal',
        'Árbol Regresión (depth=6)',
        'Naive Bayes',
        'KNN',
        'Red Neuronal (ReLU)'
    ],
    'R2': [
        0.818402,
        0.9473,
        0.807,
        0.6605,
        0.8363,
        0.8350
    ],
    'Tiempo (s)': [
        0.3,
        1.6,
        0.4,
        20.0,
        0.1,
        107
    ]
}

df = pd.DataFrame(data)

# Gráfico 1: R² por modelo
plt.figure(figsize=(10, 5))
plt.bar(df['Modelo'], df['R2'], color='skyblue')
plt.title('Comparación de R² por Modelo')
plt.ylabel('R²')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# Gráfico 2: Tiempo de ejecución por modelo
plt.figure(figsize=(10, 5))
plt.bar(df['Modelo'], df['Tiempo (s)'], color='salmon')
plt.title('Comparación de Tiempo de Ejecución por Modelo')
plt.ylabel('Tiempo (segundos)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

SVM Lineal lidera también en esta métrica, con un R² de 0.9473, indicando que es el modelo que mejor explica la variabilidad del precio de las casas.

Le siguen KNN (0.8363) y la Red Neuronal (0.8350), lo cual confirma su buen ajuste, aunque inferior al del SVM.

El rendimiento de Naive Bayes (0.6605) revela limitaciones importantes en la capacidad de modelar relaciones complejas.

Eficiencia computacional (tiempo de ejecución):

KNN fue el modelo más eficiente, con un tiempo de ejecución de 0.1s, seguido por MLR (0.3s) y el Árbol de Regresión (0.4s).

SVM Lineal, tardó 1.6s, lo cual es razonable dada su alta precisión.

En contraste, la Red Neuronal (107s) y Naive Bayes (20s) fueron los modelos más costosos computacionalmente.

# Modelo Recomendado

Considerando de manera global la precisión predictiva (MAE, RMSE), la capacidad explicativa (R²) y el tiempo de ejecución, el modelo SVM Lineal se destaca como la mejor alternativa:

- Es el más preciso,
- El que mejor explica la variabilidad del precio,
- Mantiene un tiempo de ejecución bajo (1.6s), perfectamente viable para la mayoría de aplicaciones prácticas.

# Modelos predictores de clasificacion



## Accuracy, Recall y F1-score por modelo

```python
import pandas as pd
import matplotlib.pyplot as plt

# Datos de métricas
data = {
    'Modelo': [
        'Naive Bayes',
        'KNN',
        'RNA (Tangente)',
        'Random Forest',
        'Decision Tree',
        'SVM'
    ],
    'Accuracy': [0.74519, 0.79, 0.9726, 0.8267, 0.7388, 0.9925],
    'Recall':   [0.74519, 0.79, 0.9644, 0.8301, 0.7343, 0.9924],
    'F1':       [0.74519, 0.79, 0.9643, 0.8244, 0.7345, 0.9924]
}

df = pd.DataFrame(data)

# Gráfico de métricas
df_plot = df.set_index('Modelo')
df_plot.plot(kind='bar', figsize=(12, 6))
plt.title('Comparación de Accuracy, Recall y F1 por Modelo')
plt.ylabel('Valor')
plt.ylim(0.7, 1.05)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.legend(title='Métrica')
plt.show()
```

Estas tres métricas resumen el comportamiento predictivo de cada modelo:

- SVM sobresale con un accuracy de 0.9925, acompañado de valores prácticamente idénticos en recall (0.9924) y F1-score (0.9924), lo que indica un modelo muy preciso y equilibrado.

- Le sigue RNA con función tangente, que alcanza 0.9726 de accuracy, también con alta consistencia entre recall y F1-score (ambos ≈0.964).

- Random Forest (profundidad 15) muestra un buen balance con 0.8267 de accuracy, destacando frente a modelos más simples como KNN (0.79) y Naive Bayes (0.745).

- Decision Tree fue el de menor rendimiento (0.7388), aunque aún aceptable en problemas no lineales.

## Tiempo de ejecuciónTiempo de ejecución

```python
# Datos de tiempos (en segundos)
tiempos = {
    'Modelo': [
        'Naive Bayes',
        'KNN',
        'RNA (Tangente)',
        'Random Forest',
        'Decision Tree',
        'SVM'
    ],
    'Tiempo (s)': [0.1, 3.7, 0.3, 0.1505, 1.33, 0.1]
}

df_time = pd.DataFrame(tiempos)

# Gráfico de tiempos
plt.figure(figsize=(10, 5))
plt.bar(df_time['Modelo'], df_time['Tiempo (s)'], color='salmon')
plt.title('Comparación de Tiempos de Ejecución por Modelo')
plt.ylabel('Tiempo (segundos)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

KNN es el más lento, ya que su clasificación requiere calcular distancias en tiempo real para cada punto.
En contraste, SVM y Naive Bayes fueron los más rápidos (0.1s), seguidos muy de cerca por Random Forest (0.15s) y RNA (0.3s). Decision Tree fue algo más lento (1.33s), posiblemente por poda o complejidad estructural.

## Conclusión

- SVM es claramente el mejor modelo de clasificación en este análisis, logrando un equilibrio sobresaliente entre precisión y velocidad de ejecución.
- Si se busca una alternativa con menor costo computacional y rendimiento alto, RNA (Tangente) y Random Forest también son opciones recomendables.
- KNN, Naive Bayes y Árbol de Decisión pueden ser útiles como modelos base o para interpretabilidad, pero presentan limitaciones notables en precisión o eficiencia.
